## 🎯 Objective
You are a Data Scientist at a University. Your task is to **predict the lifestyle profiles** based on behaviourial metrics.  
You will test with **several classification techniques** and compare their performance.

## 📂 Dataset Description
The dataset contains measurements for different students, including:

- `student_id`: A unique identifier for each student (used only for reference and submission, not for learning).
- `study_hours_per_week`: Total number of hours the student spends studying in a typical week (numeric).
- `exercise_hours_per_week`: Total number of hours the student spends doing physical exercise in a typical week (numeric).
- `sleep_hours_per_night`: Average number of hours the student sleeps per night (numeric, usually between 4 and 10).
- `screen_hours_per_day`: Average number of hours per day the student spends on screens outside of school work (phone, computer, TV, etc.) (numeric).
- `favorite_subject`: The student’s favorite school subject (categorical: for example Math, Science, Language, Arts, Other).
- `enjoys_group_work`: Whether the student likes working in groups for school projects (categorical: Yes or No).
- `lunch_at_school`: Whether the student usually has lunch at school (categorical: Yes or No).
  
- **Target**: `lifestyle_profile` The label you want to predict: the student’s lifestyle profile, based on their habits (categorical, with four classes such as `Balanced`, `Academic_Focused`, `Sport_Focused`, `Screen_Heavy`).
