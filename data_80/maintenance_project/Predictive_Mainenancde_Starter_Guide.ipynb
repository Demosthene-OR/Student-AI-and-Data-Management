**Two approaches:**

2. **Classification target**: Binary (will fail in next N cycles/hours?)

```python
# TODO: PUMPS - Prepare data for modeling
# Step 1: Remove NaN values
# Step 2: Select feature columns (exclude machine_id, cycle, rul_cycles)
# Step 3: Create regression target (rul_cycles)
# Step 4: Create classification target (failure in next 30 cycles?)
# Step 5: Normalize features with StandardScaler
# Step 6: Split train/test (80-20)

failure_horizon_pump = 30

# Select feature columns (all sensors + engineered features, but not IDs or targets)
feature_cols_pump = [col for col in pump_df_fe.columns 
                     if col not in ['engine_id', 'cycle', 'rul_cycles']]

print(f"Number of features: {len(feature_cols_pump)}")
print(f"Features: {feature_cols_pump}")

# YOUR CODE HERE
# Step 1: Clean data
pump_clean = pump_df_fe.dropna()

# Step 2: Extract features and targets
X_pump = pump_clean[feature_cols_pump].values
y_pump_rul = pump_clean['rul_cycles'].values
y_pump_clf = (pump_clean['rul_cycles'] <= failure_horizon_pump).astype(int).values

print(f"\nX shape: {X_pump.shape}")
print(f"y_rul shape: {y_pump_rul.shape}")
print(f"Classification balance: {y_pump_clf.sum()} failures / {len(y_pump_clf)} total")

# Step 3: Normalize
scaler_pump = StandardScaler()
X_pump_scaled = scaler_pump.fit_transform(X_pump)

# Step 4: Split train/test
X_pump_train, X_pump_test, y_pump_rul_train, y_pump_rul_test = train_test_split(
    X_pump_scaled, y_pump_rul, test_size=0.2, random_state=42
)
_, _, y_pump_clf_train, y_pump_clf_test = train_test_split(
    X_pump_scaled, y_pump_clf, test_size=0.2, random_state=42
)

print(f"\n✓ Train set: {X_pump_train.shape[0]} samples")
print(f"✓ Test set: {X_pump_test.shape[0]} samples")
```
